```python
important_columns = [

    "year",
    "make",
    "model",
    "VClass",
    "cylinders",
    "displ",
    "drive",
    "fuelType",
    "city08",
    "highway08",
    "comb08",
    "fuelCost08"
]

df[important_columns].head()
```

## 2. Data Cleaning

### Check Missing Values

```python
df[important_columns].isnull().sum()
```

### Create Working Dataset

## we take the columns we selected and make a separate working copy called vehicles.

```python
vehicles = df[important_columns].copy()

vehicles.head()
```

## 3. Filter for SUVs

### Check Vehicle Classes

```python
vehicles["VClass"].value_counts()
```

### Create SUV Dataset

```python
suv = vehicles[vehicles["VClass"].str.contains("Sport Utility Vehicle",case=False, na=False)].copy()

suv.head()
```

```python
suv.shape
```

```python
suv["VClass"].value_counts()
```

## oldest and newest model years in suv data?

```python
suv["year"].min(), suv["year"].max()
```

## 4. Select Analysis Period

For this analysis, we will focus on SUVs from model years 2015 through 2025.

```python
suv_recent = suv[
    (suv["year"] >= 2015)&
    (suv["year"] <= 2025)


].copy()

suv_recent.shape
```

## 5. Explore SUV Makes and Models

```python
suv_recent["make"].value_counts()
```

## check for Toyota 

```python
suv_recent[suv_recent["make"] == "Toyota"]["make"].value_counts()
```

## Which Toyota models are those 257 records?

```python
suv_recent[suv_recent["make"] == "Toyota"]["model"].value_counts()
```

## 6. Analyze Toyota Sequoia

```python
sequoia = suv_recent[(suv_recent["make"] == "Toyota") & (suv_recent["model"].str.contains("Sequoia", case = False, na = False))].copy()

sequoia.shape
```

```python
sequoia[["year", "model", "cylinders", "displ", "drive",
         "city08", "highway08", "comb08", "fuelCost08"]]
```

## 7. Compare Sequoia by Model Year

Compare Sequoia engine configuration, fuel economy, and estimated fuel cost across model years.

```python
sequoia_sorted = sequoia.sort_values("year")
sequoia_sorted[[
    "year",
    "model",
    "cylinders",
    "displ",
    "city08",
    "highway08",
    "comb08",
    "fuelCost08"
]]
```

## 8. Average Fuel Economy by Year

```python
sequoia_yearly = sequoia.groupby("year")[["comb08","fuelCost08"]].mean()

sequoia_yearly
```

## 9. Visualize Sequoia Fuel Economy Trend

```python
sequoia_yearly["comb08"].plot(
    kind="line",
    marker="o",
    title="Toyota Sequoia Average Combined MPG by Year"
)
```

## 10. Calculate MPG Improvement After 2023 Redesign

```python
mpg_2022 = sequoia_yearly.loc[2022, "comb08"]
mpg_2023 = sequoia_yearly.loc[2023, "comb08"]

mpg_improvement = ((mpg_2023 - mpg_2022)/mpg_2022)*100

round(mpg_improvement,2)
```

## 11. Calculate Fuel Cost Reduction After 2023 Redesign

```python
fuelcost_2022 = sequoia_yearly.loc[2022,"fuelCost08"]
fuelcost_2023 = sequoia_yearly.loc[2023,"fuelCost08"]

fuelcost_reduction = ((fuelcost_2022 - fuelcost_2023)/fuelcost_2022)*100

fuelcost_reduction
```

## 12. Identify Full-Size SUV Competitors

```python
suv_recent[
    suv_recent["make"].isin(["Toyota", "Chevrolet", "Ford", "GMC"])
][["make", "model"]].drop_duplicates().sort_values(["make", "model"])
```

## search specifically for the four models we care about.
## Find models containing Sequoia OR Tahoe OR Expedition OR Yukon.

```python
suv_recent[
    suv_recent["model"].str.contains(
        "Sequoia|Tahoe|Expedition|Yukon",
        case=False,
        na=False
    )
][["make", "model"]].drop_duplicates().sort_values(["make", "model"])
```

## 13. Create Full-Size SUV Competitor Dataset

## Keep a row only when its model is exactly one of these models.
## For now, let's compare the standard Sequoia, Tahoe, Expedition, and Yukon and exclude things like Expedition MAX, Yukon XL, and FFV variants.

```python
competitors = suv_recent[
    suv_recent["model"].isin([
        "Sequoia 2WD",
        "Sequoia 4WD",
        "Tahoe 2WD",
        "Tahoe 4WD",
        "Expedition 2WD",
        "Expedition 4WD",
        "Yukon 2WD",
        "Yukon 4WD"
    ])
].copy()

competitors[["make", "model"]].value_counts()
```

## 14. Compare Average Fuel Economy by Manufacturer

```python
competitor_mpg = competitors.groupby("make")["comb08"].mean()

competitor_mpg
```

## 15. Compare Competitor MPG by Year

```python
competitor_yearly = (
    competitors
    .groupby(["year", "make"])["comb08"]
    .mean()
    .reset_index()
)

competitor_yearly
```

## 16. Prepare Competitor Data for Visualization

```python
competitor_pivot = competitor_yearly.pivot(
    index="year",
    columns="make",
    values="comb08"
)

competitor_pivot
```

## Check which years each model name appears

```python
suv_recent[
    suv_recent["model"].str.contains("Tahoe|Yukon", case=False, na=False)
][["year", "make", "model"]].drop_duplicates().sort_values(
    ["year", "make", "model"]
)
```

## 18. Inspect Tahoe Naming Across Years

```python
suv_recent[
    suv_recent["model"].str.contains("Tahoe", case=False, na=False)
][["year", "model"]].drop_duplicates().sort_values(["year", "model"])
```

## 18. Clean Full-Size SUV Competitor Selection

```python
competitors = suv_recent[
    (
        suv_recent["model"].str.contains("Sequoia", case=False, na=False)
    )
    |
    (
        suv_recent["model"].str.contains("Tahoe", case=False, na=False)
    )
    |
    (
        suv_recent["model"].str.contains("Expedition", case=False, na=False)
        & ~suv_recent["model"].str.contains("MAX|EL", case=False, na=False)
    )
    |
    (
        suv_recent["model"].str.contains("Yukon", case=False, na=False)
        & ~suv_recent["model"].str.contains("XL", case=False, na=False)
    )
].copy()
```

```python
competitors[["make", "model"]].drop_duplicates().sort_values(
    ["make", "model"]
)
```

## 19. Recalculate Competitor MPG by Year

```python
competitor_yearly = (
    competitors
    .groupby(["year", "make"])["comb08"]
    .mean()
    .reset_index()
)

competitor_yearly
```

## clear comparison chart for Toyota, Chevrolet, Ford, and GMC across the years.

```python
competitor_pivot = competitor_yearly.pivot(
    index="year",
    columns="make",
    values="comb08"
)

competitor_pivot.plot(
    kind="line",
    marker="o",
    title="Average Combined MPG: Full-Size SUV Competitors"
)
```
